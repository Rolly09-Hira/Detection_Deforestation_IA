```python
#structurer les dossier 
import shutil
from pathlib import Path
from sklearn.model_selection import train_test_split
```

```python
#dossier des données original et sortie
RAW_PATH = Path("/home/rolls/M1/IA/DataScience/Projets/Detection_Deforestation_IA/data/raw")
TRAIN_PATH = RAW_PATH.parent / "train"
TEST_PATH = RAW_PATH.parent / "test"
VALIDATION_PATH = RAW_PATH.parent / "validation"
```

```python
#creation auto des dossier
for dossier in [TRAIN_PATH, TEST_PATH, VALIDATION_PATH]:
    dossier.mkdir(parents=True, exist_ok=True)
#creation des sous dossier pour chaque classe
for classe in ["forest", "non_forest"]:
    (TRAIN_PATH / classe).mkdir(exist_ok = True)
    (TEST_PATH / classe).mkdir(exist_ok = True)
    (VALIDATION_PATH / classe).mkdir(exist_ok = True)
print("Structure creer avec succes")
```

```python
#Fonction de division
def obtenir_img(dossier):
    return list(dossier.rglob("*.png"))
```

```python
#recuperer les images
forest_images = obtenir_img(RAW_PATH / "forest")
non_forest_images = obtenir_img(RAW_PATH / "non_forest")

print(f"Forest : {len(forest_images)}")
print(f"Non Forest : {len(non_forest_images)}")
```

```python
#Diviser les dataset
##copier les images:
def copier_images(images, dossier_destination):
    for image in images:
        shutil.copy2(image, dossier_destination / image.name)
# -------- FOREST --------
forest_train, forest_temp = train_test_split(
    forest_images,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

forest_validation, forest_test = train_test_split(
    forest_temp,
    test_size=0.50,
    random_state=42,
    shuffle=True
)

# -------- NON-FOREST --------
non_train, non_temp = train_test_split(
    non_forest_images,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

non_validation, non_test = train_test_split(
    non_temp,
    test_size=0.50,
    random_state=42,
    shuffle=True
)
print("=== Forest ===")
print("Train :", len(forest_train))
print("Validation :", len(forest_validation))
print("Test :", len(forest_test))

print("\n=== Non-Forest ===")
print("Train :", len(non_train))
print("Validation :", len(non_validation))
print("Test :", len(non_test))
```

```python
#copier les image

#forest
copier_images(forest_train, TRAIN_PATH / "forest")
copier_images(forest_validation, VALIDATION_PATH / "forest")
copier_images(forest_test, TEST_PATH / "forest")

#non forest
copier_images(non_train, TRAIN_PATH / "non_forest")
copier_images(non_validation, VALIDATION_PATH / "non_forest")
copier_images(non_test, TEST_PATH / "non_forest")
print("Copie terminer")
```

```python
from pathlib import Path

def compter_images(dossier):
    return len(list(Path(dossier).rglob("*.png")))


print("===== TRAIN =====")
print("Forest :", compter_images(TRAIN_PATH / "forest"))
print("Non-Forest :", compter_images(TRAIN_PATH / "non_forest"))


print("\n===== VALIDATION =====")
print("Forest :", compter_images(VALIDATION_PATH / "forest"))
print("Non-Forest :", compter_images(VALIDATION_PATH / "non_forest"))


print("\n===== TEST =====")
print("Forest :", compter_images(TEST_PATH / "forest"))
print("Non-Forest :", compter_images(TEST_PATH / "non_forest"))
```
